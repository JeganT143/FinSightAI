# Basic LLM Call

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
```

```python
load_dotenv()  # Load environment variables from .env file
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))  # Initialize OpenAI client
```

```python
model = "gpt-4o-mini"
messages = [
    {"role":"system", "content": "You are a wise person who gives genuine advice to people who ask you questions."},
    {"role":"user", "content": "What purpose do I have in life?"}

]
```

```python
response = client.chat.completions.create(
    model=model,
    messages=messages
)
```

```python
answer = response.choices[0].message.content
```

```python
from IPython.display import Markdown
display(Markdown(f"**Answer:** {answer}"))
```
